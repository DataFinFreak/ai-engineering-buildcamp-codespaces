# Stage 1 - Baseline

**Task for the user** (type it yourself, do not copy from the course notebook):

- import `GithubRepositoryDataReader` from `gitsource`
- owner `evidentlyai`, repo `docs`, allowed extensions `md` and `mdx`
- call `.read()` into a variable `files`

Then report:

1. `len(files)`
2. `type(files[0])`
3. `files[0].filename`
4. A guess, **before running**: what is `.read()` doing while it hangs?

Gotchas worth hitting once:
- `allowed_extensions` takes **no dots** - `{"md", "mdx"}`, not `{".md"}`. Dots return zero files with no error.
- It is a **set**, not a list.
