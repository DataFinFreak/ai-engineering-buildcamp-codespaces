# Smoke test

Confirms the environment is wired up: `.env` loads, the OpenAI client authenticates, and a completion comes back.

```python
import os
from dotenv import load_dotenv

# .env lives at the repo root, two levels up from this notebook
load_dotenv("../../.env")

key = os.environ.get("OPENAI_API_KEY")
# never print the key itself - this repo is public
print("key loaded:", bool(key), "| length:", len(key) if key else 0)
```

```python
from openai import OpenAI

client = OpenAI()

resp = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[{"role": "user", "content": "Reply with exactly: environment ok"}],
)
print(resp.choices[0].message.content)
```

```python
print("model:", resp.model)
print("tokens used:", resp.usage.total_tokens)
```
